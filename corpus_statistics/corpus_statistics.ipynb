```python
import os
from estnltk.converters import json_to_text
```

### Importing corpus data

```python
TimeML_corpus_path = 'C:/Users/liivas/Documents/Magistritoo/andmetootlus/corpus_preprocessing/EstTimeML_corpus_json/'
temp_fact_corpus_path = 'C:/Users/liivas/Documents/Magistritoo/andmetootlus/corpus_preprocessing/temporal_facts_corpus_json/'
```

```python
TimeML_texts = []

for filename in os.listdir(TimeML_corpus_path):
    text_obj = json_to_text(file=TimeML_corpus_path + filename)
    TimeML_texts.append(text_obj)
```

```python
temp_fact_texts = []

for filename in os.listdir(temp_fact_corpus_path):
    text_obj = json_to_text(file=temp_fact_corpus_path + filename)
    temp_fact_texts.append(text_obj)
```

### Corpora size in words

```python
# The size of both corpora in words is calculated by the sum of EstNLTK words

n_words_TimeML = 0
n_words_temp_fact = 0

for text in TimeML_texts:
    n_words_TimeML+=len(text.words)
    
for text in temp_fact_texts:
    n_words_temp_fact+=len(text.words)

print(f"TimeML corpus size in words: {n_words_TimeML}")
print(f"Temporal facts corpus size in words: {n_words_temp_fact}")
```

### Number of events

```python
# Event layers of TimeML corpus and temporal facts corpus differ.
# In order to count the number of events, we will count event phrases instead of singular event words in TimeML corpus.
# The reason is that event phrases represent event entities more accurately 

n_events_TimeML = 0
n_events_temp_fact = 0

for text in TimeML_texts:
    n_events_TimeML+=len(text.gold_event_phrases)
    
for text in temp_fact_texts:
    n_events_temp_fact+=len(text.events)

print(f"TimeML corpus number of events: {n_events_TimeML}")
print(f"Temporal facts corpus number of events: {n_events_temp_fact}")
```

### Number of multiword events

```python
# Event layers of TimeML corpus and temporal facts corpus differ.
# In order to count the number of events, we will count event phrases instead of singular event words in TimeML corpus.
# The reason is that event phrases represent event entities more accurately 

n_multiw_events_TimeML = 0
n_multiw_events_temp_fact = 0

for text in TimeML_texts:
    for event in text.gold_event_phrases:
        if len(event) > 1:
            n_multiw_events_TimeML+=1
    
for text in temp_fact_texts:
    for event in text.events:
        if len(event) > 1:
            n_multiw_events_temp_fact+=1

print(f"TimeML corpus number of multiword events: {n_multiw_events_TimeML}, {(n_multiw_events_TimeML/n_events_TimeML)*100} % of all TimeML events")
print(f"Temporal facts corpus number of multiword events: {n_multiw_events_temp_fact}, {(n_multiw_events_temp_fact/n_events_temp_fact)*100} % of all tempf events")
```

### Number of timexes

```python
# Timex layers of TimeML corpus and temporal facts corpus differ.
# In order to count the number of timexes, we will count timex phrases instead of singular timex words in TimeML corpus.
# The reason is that timex phrases represent timex entities more accurately.

n_timexes_TimeML = 0
n_timexes_temp_fact = 0

for text in TimeML_texts:
    n_timexes_TimeML+=len(text.gold_timex_phrases)
    
for text in temp_fact_texts:
    n_timexes_temp_fact+=len(text.timexes)

print(f"TimeML corpus number of timexes: {n_timexes_TimeML}")
print(f"Temporal facts corpus number of timexes: {n_timexes_temp_fact}")
```

### Number of entities

```python
# Entity annotation exists only in temporal facts corpus

n_entities_temp_fact = 0

for text in temp_fact_texts:
    n_entities_temp_fact+=len(text.entities)

print(f"Temporal facts corpus number of entities: {n_entities_temp_fact}")
```

### Number of event-timex TLINKs

```python
# TLINK relations between events and timexes in both corpora.
# Additionally, TLINK relations between all events inside sentences and between main events of two consecutive sentences
# exist in TimeML corpus. However, these relations will not be analysed initially
# so they will not be included in current statistics.

n_e_t_tlinks_TimeML = 0
n_e_t_tlinks_temp_fact = 0
n_ent_t_tlinks_temp_fact = 0


for text in TimeML_texts:
    n_e_t_tlinks_TimeML+=len(text['event_timex_tlinks'])
    
for text in temp_fact_texts:
    for tlink in text['tlinks']:
        ev_word_spans = [text.words.get(span) for span in tlink.a_text.base_span]
        is_event = True
        # jätame entiteedid välja
        for span in ev_word_spans:
            for ent in text.entities:
                if span in ent:
                    is_event = False
                    break
        if not is_event:
            #print(tlink.a_text.base_span)
            #n_e_t_tlinks_temp_fact+=len(text['tlinks'])
            n_ent_t_tlinks_temp_fact+=1
        else:
            n_e_t_tlinks_temp_fact+=1

print(f"TimeML corpus number of event-timex TLINKs: {n_e_t_tlinks_TimeML}")
print(f"Temporal facts corpus number of event-timex TLINKs: {n_e_t_tlinks_temp_fact}")
print(f"Temporal facts corpus number of entity-timex TLINKs: {n_ent_t_tlinks_temp_fact}")
```

### Number of event-dct TLINKs

```python
# TLINKs between events and DCT exist only in TimeML corpus

n_e_dct_tlinks_TimeML = 0

for text in TimeML_texts:
    n_e_dct_tlinks_TimeML+=len(text['event_dct_tlinks'])
    
print(f"TimeML corpus number of event-DCT TLINKs: {n_e_dct_tlinks_TimeML}")
```

### Number of event arguments

```python
# Relations between main events and their immediate arguments exist only in temporal facts corpus.

n_e_rels_temp_fact = 0

for text in temp_fact_texts:
    n_e_rels_temp_fact+=len(text['event_arguments'])
    
print(f"Temporal facts corpus number of event argument relations: {n_e_rels_temp_fact}")
```

### TLINKs in TimeML corpus

```python
def get_tlink_data(text_objects, tlink_layer):
    tlink_data = {'SIMULTANEOUS': 0, 'BEFORE-OR-OVERLAP': 0, 'OVERLAP-OR-AFTER': 0, 'BEFORE': 0, 'AFTER': 0, 'INCLUDES': 0, 'IS_INCLUDED': 0, 'VAGUE': 0, 'IDENTITY': 0}

    for text in text_objects:
        for tlink in text[tlink_layer]:
            tlink_data[tlink['rel_type']]+=1
            
    return tlink_data
```

```python
event_timex_tlinks = get_tlink_data(TimeML_texts, 'event_timex_tlinks')

for rel_type_data in event_timex_tlinks:
    print(rel_type_data, event_timex_tlinks[rel_type_data])
```

```python
event_dct_tlinks = get_tlink_data(TimeML_texts, 'event_dct_tlinks')

total = 0
for rel_type_data in event_dct_tlinks:
    total+=event_dct_tlinks[rel_type_data]
    print(rel_type_data, event_dct_tlinks[rel_type_data])
    
print()
print(total)
```

### Event classes

#### TimeML

```python
def get_event_class_data(text_objects, event_layer):
    event_class_data = {'REPORTING': 0, 'PERCEPTION': 0, 'ASPECTUAL': 0, 'I_ACTION': 0, 'I_STATE':0, 'STATE': 0, 
                      'MODAL': 0, 'OCCURRENCE': 0}

    for text in text_objects:
        for event in text[event_layer]:
            event_class_data[event['event_class'][0]]+=1
            
    return event_class_data
```

```python
event_classes = get_event_class_data(TimeML_texts, 'gold_event_phrases')

total = 0
for ev_class in event_classes:
    total+=event_classes[ev_class]
    print(ev_class, event_classes[ev_class])

print()
print(total)
```

#### Temporal facts

```python
def get_event_class_data2(text_objects, event_layer):
    event_class_data = {'REPORTING': 0, 'PERCEPTION': 0, 'ASPECTUAL': 0, 'I_ACTION': 0, 'I_STATE':0, 'STATE': 0, 
                      'MODAL': 0, 'OCCURRENCE': 0}

    for text in text_objects:
        for event in text[event_layer]:
            event_class_data[event['class']]+=1
            
    return event_class_data
```

```python
event_classes2 = get_event_class_data2(temp_fact_texts, 'events')

total = 0
for ev_class in event_classes2:
    total+=event_classes2[ev_class]
    print(ev_class, event_classes2[ev_class])

print()
print(total)
```

### Event class examples in temporal facts corpus

```python
event_class_ex = {'REPORTING': [], 'PERCEPTION': [], 'ASPECTUAL': [], 'I_ACTION': [], 'I_STATE': [], 'STATE': [], 
                      'MODAL': [], 'OCCURRENCE': []}

for text in temp_fact_texts:
    for event in text.events:
        if len(event_class_ex[event['class']])==0:
            for sentence in text.sentences:
                if event[0] in sentence.words:
                    event_class_ex[event['class']].append(event.text)
                    event_class_ex[event['class']].append(' '.join(sentence.text))
```

```python
for ex in event_class_ex:
    print(ex, event_class_ex[ex][0])
    print(event_class_ex[ex][1])
    print()
```

### Event durations

```python
def get_event_dur_data(text_objects, event_layer):
    event_dur_data = {'instant': 0, 'seconds': 0, 'minutes': 0, 'hours': 0, 'days':0, 'weeks': 0, 
                      'months': 0, 'years': 0, 'centuries': 0, 'forever': 0}

    for text in text_objects:
        for event in text[event_layer]:
            event_dur_data[event['duration']]+=1
            
    return event_dur_data
```

```python
# event durations are annotated only in temporal facts corpus
event_durations = get_event_dur_data(temp_fact_texts, 'events')

total = 0
for dur in event_durations:
    total+=event_durations[dur]
    print(dur, event_durations[dur])

print()
print(total)
```

### TLINK examples in temporal facts corpus

```python
tlink_ex = {'SIMULTANEOUS': [], 'BEFORE': [], 'AFTER': [], 'INCLUDES': [], 'IS_INCLUDED': [], 'VAGUE': []}

for text in temp_fact_texts:
    for rel in text['tlinks']:
        for sentence in text.sentences:
            for i in range(len(sentence.words)):
                if sentence.words[i].base_span == rel['a_text'].base_span[0]:
                    lst = []
                    lst.append(rel['a_text'])
                    lst.append(rel['b_text'])
                    lst.append(' '.join(sentence.text))
                    tlink_ex[rel['rel_type'][0]].append(lst)
```
