## Reading temporal facts corpus

```python
import os
from corpus_methods.convert_BRAT_to_estnltk_json_updated import import_from_brat_folder
from corpus_methods.file_operations import *
```

### Creating Text-objects with layers

```python
text_objects = import_from_brat_folder('initial_data/temporal_facts_corpus')
```

### Saving processed data

```python
# -- new target directory for temporal facts corpus articles in JSON format
json_dir = 'processed_data/temporal_facts_corpus_json'

if not os.path.isdir(json_dir):
    os.mkdir(json_dir)

# -- target folder path
json_dir_path = 'processed_data/temporal_facts_corpus_json/'

n_saved_articles = 0
for text_obj in text_objects:
    print(text_obj.meta['filename'])
    
    # save Text-object in JSON format
    save_Text_to_json(json_dir_path, text_obj)
    n_saved_articles+=1
    
print()
print(f"Saved {n_saved_articles} articles in JSON-format.")
```
