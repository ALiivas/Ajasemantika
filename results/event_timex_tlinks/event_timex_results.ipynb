## Event-timex TLINKs identification results

```python
import os
from collections import OrderedDict
```

#### Importing and organizing model training results

```python
estbert_ev_timex_path = "estbert_ev_timex_res/estbert_ev_timex_accuracy.txt"
estbert_ev_timex_agg_path = "estbert_ev_timex_agg_res/estbert_ev_timex_agg_accuracy.txt"

estroberta_ev_timex_path = "estroberta_ev_timex_res/estroberta_ev_timex_accuracy.txt"
estroberta_ev_timex_agg_path = "estroberta_ev_timex_agg_res/estroberta_ev_timex_agg_accuracy.txt"
```

```python
estbert_ev_timex_res = {}
estbert_ev_timex_agg_res = {}

estroberta_ev_timex_res = {}
estroberta_ev_timex_agg_res = {}

with open(f"{estbert_ev_timex_path}") as f:
    for line in f:
       (key, val) = line.split(": ")
       estbert_ev_timex_res[key] = float(val)
       
with open(f"{estbert_ev_timex_agg_path}") as f:
    for line in f:
       (key, val) = line.split(": ")
       estbert_ev_timex_agg_res[key] = float(val)
       
       
with open(f"{estroberta_ev_timex_path}") as f:
    for line in f:
       (key, val) = line.split(": ")
       estroberta_ev_timex_res[key] = float(val)
       
with open(f"{estroberta_ev_timex_agg_path}") as f:
    for line in f:
       (key, val) = line.split(": ")
       estroberta_ev_timex_agg_res[key] = float(val)
```

### EstBERT results

#### All TLINK classes

```python
# estbert sündmus-ajaväljend
estbert_ev_timex_res_desc = OrderedDict(sorted(estbert_ev_timex_res.items(), 
                                  key=lambda kv: kv[1], reverse=True))

# sorted
for key, value in dict(estbert_ev_timex_res_desc).items():
    print(key, value)
```

#### TLINKs aggregated into 4 classes

```python
# estbert sündmus-ajaväljend agregeeritult
estbert_ev_timex_agg_res_desc = OrderedDict(sorted(estbert_ev_timex_agg_res.items(), 
                                  key=lambda kv: kv[1], reverse=True))

# sorted
for key, value in dict(estbert_ev_timex_agg_res_desc).items():
    print(key, value)
```
