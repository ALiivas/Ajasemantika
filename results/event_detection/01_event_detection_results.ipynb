## Event detection results

```python
import pandas as pd
import os
import json
```

```python
ESTBERT_RESULTS_DIR = 'results/EstBERT'
ESTROBERTA_RESULTS_DIR = 'results/Est-RoBERTa'
```

#### Methods

```python
def get_train_res(file, metrics):
    data = None
    with open(file) as f:
        data = json.load(f)
    res_dict = dict()
    for metric in metrics:
        res_dict[f'train_{metric}'] = round(data[metric], 2)
    return res_dict

def get_test_res(file, eval_method, metrics):
    data = None
    with open(file) as f:
        data = json.load(f)
    res_dict = dict()
    for metric in metrics:
        res_dict[f'final_test_{metric}'] = round(data[0][eval_method][metric], 2)
    return res_dict

def get_pred_res(file, eval_method, metrics):
    data = None
    with open(file) as f:
        data = json.load(f)
    res_dict = dict()
    for metric in metrics:
        res_dict[f'pred_{metric}'] = round(data[0][eval_method][metric], 2)
    return res_dict

def fill_res_dict(res_dict, model_cat, path, event_type):
    for filename in os.listdir(path):
        if 'best_eval' in filename:
            res = get_train_res(os.path.join(path, filename), ['eval_precision', 'eval_recall', 'eval_f1'])
            res_dict[model_cat][f'train_eval_{event_type}'] = res
        elif 'best_final_test' in filename:
            res = get_test_res(os.path.join(path, filename), 'strict', ['precision', 'recall', 'f1'])
            res_dict[model_cat][f'final_test_{event_type}'] = res
        elif 'pred_tempf' in filename:
            res = get_pred_res(os.path.join(path, filename), 'strict', ['recall'])
            if 'rkogu' in filename:
                res_dict[model_cat][f'pred_tempf_{event_type}_rkogu'] = res
            elif 'horisont' in filename:
                res_dict[model_cat][f'pred_tempf_{event_type}_horisont'] = res
            elif 'news' in filename:
                res_dict[model_cat][f'pred_tempf_{event_type}_news'] = res
```

```python
# EstBERTi tulemused sündmuste kõigil sõnadel ja ainult peasõnadel
estbert_res = {'EstBERT_no_classes': {}, 'EstBERT_w_classes': {}, 'EstBERT_w_agg_classes': {}, 'EstBERT_w_agg_classes_no_modal': {}}

# ilma klassideta
fill_res_dict(estbert_res, 'EstBERT_no_classes', f"{ESTBERT_RESULTS_DIR}/events/all", 'all')
fill_res_dict(estbert_res, 'EstBERT_no_classes', f"{ESTBERT_RESULTS_DIR}/events/main", 'main')

# klassidega
fill_res_dict(estbert_res, 'EstBERT_w_classes', f"{ESTBERT_RESULTS_DIR}/events_w_class/all", 'all')
fill_res_dict(estbert_res, 'EstBERT_w_classes', f"{ESTBERT_RESULTS_DIR}/events_w_class/main", 'main')

# agregeeritud klassidega (koos eraldi modaalverbide kategooriaga)
fill_res_dict(estbert_res, 'EstBERT_w_agg_classes', f"{ESTBERT_RESULTS_DIR}/events_w_agg_class/all", 'all')
fill_res_dict(estbert_res, 'EstBERT_w_agg_classes', f"{ESTBERT_RESULTS_DIR}/events_w_agg_class/main", 'main')

# agregeeritud klassidega (ilma eraldi modaalverbide kategooriata)
fill_res_dict(estbert_res, 'EstBERT_w_agg_classes_no_modal', f"{ESTBERT_RESULTS_DIR}/events_w_agg_class_no_modal/all", 'all')
fill_res_dict(estbert_res, 'EstBERT_w_agg_classes_no_modal', f"{ESTBERT_RESULTS_DIR}/events_w_agg_class_no_modal/main", 'main')
```
