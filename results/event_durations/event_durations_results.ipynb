## Event duration assingment results on final test set

```python
import os
from collections import OrderedDict
```

#### Importing and organizing model training results

```python
estbert_ev_dur_path = "estbert_ev_durations_res/estbert_ev_dur_accuracy.txt"
estbert_ev_dur_agg_path = "estbert_ev_durations_agg_res/estbert_ev_dur_accuracy.txt"
estbert_ev_dur_agg_binary_path = "estbert_ev_durations_agg_binary_res/estbert_ev_dur_accuracy.txt"

estroberta_ev_dur_path = "estroberta_ev_durations_res/estroberta_ev_dur_accuracy.txt"
estroberta_ev_dur_agg_path = "estroberta_ev_durations_agg_res/estroberta_ev_dur_accuracy.txt"
estroberta_ev_dur_agg_binary_path = "estroberta_ev_durations_agg_binary_res/estroberta_ev_dur_accuracy.txt"
```

```python
estbert_ev_dur_res = {}
estbert_ev_dur_agg_res = {}
estbert_ev_dur_agg_bin_res = {}

estroberta_ev_dur_res = {}
estroberta_ev_dur_agg_res = {}
estroberta_ev_dur_agg_bin_res = {}

with open(f"{estbert_ev_dur_path}") as f:
    for line in f:
       (key, val) = line.split(": ")
       estbert_ev_dur_res[key] = float(val)
       
with open(f"{estbert_ev_dur_agg_path}") as f:
    for line in f:
       (key, val) = line.split(": ")
       estbert_ev_dur_agg_res[key] = float(val)
       
with open(f"{estbert_ev_dur_agg_binary_path}") as f:
    for line in f:
       (key, val) = line.split(": ")
       estbert_ev_dur_agg_bin_res[key] = float(val)
       
with open(f"{estroberta_ev_dur_path}") as f:
    for line in f:
       (key, val) = line.split(": ")
       estroberta_ev_dur_res[key] = float(val)
       
with open(f"{estroberta_ev_dur_agg_path}") as f:
    for line in f:
       (key, val) = line.split(": ")
       estroberta_ev_dur_agg_res[key] = float(val)

with open(f"{estroberta_ev_dur_agg_binary_path}") as f:
    for line in f:
       (key, val) = line.split(": ")
       estroberta_ev_dur_agg_bin_res[key] = float(val)
```

### EstBERT results

#### All duration classes

```python
estbert_ev_dur_res_desc = OrderedDict(sorted(estbert_ev_dur_res.items(), 
                                  key=lambda kv: kv[1], reverse=True))

#print(dict(estbert_ev_dur_res_desc))
for key, value in dict(estbert_ev_dur_res_desc).items():
    print(key, value)
```

#### Durations aggregated into 3 classes

```python
estbert_ev_dur_agg_res_desc = OrderedDict(sorted(estbert_ev_dur_agg_res.items(), 
                                  key=lambda kv: kv[1], reverse=True))

#print(dict(estbert_ev_dur_agg_res_desc))
for key, value in dict(estbert_ev_dur_agg_res_desc).items():
    print(key, value)
```

#### Durations aggregated into 2 classes

```python
estbert_ev_dur_agg_bin_res_desc = OrderedDict(sorted(estbert_ev_dur_agg_bin_res.items(), 
                                  key=lambda kv: kv[1], reverse=True))

#print(dict(estbert_ev_dur_agg_bin_res_desc))
for key, value in dict(estbert_ev_dur_agg_bin_res_desc).items():
    print(key, value)
```
